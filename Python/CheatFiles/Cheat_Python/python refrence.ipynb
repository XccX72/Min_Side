## Useful functions (generelle funksjoner)
Kort tekst: Eksempler på vanlige innebygde funksjoner og string-metoder.

```
# help() viser dokumentasjon for et objekt eller funksjon
help(print)

# \n gjør at man printer på neste linje
print("jeg elser apekatt \neller gjør jeg det???")

# .islower(), .isupper() finner ut om en streng er i små eller store bokstaver
a = ("jeg heter sigma")
b = a.islower()
print("islower:", b)

# Digit / alpha / space / alnum
s = "abc123"
print("isdigit:", s.isdigit())
print("isalpha:", s.isalpha())
print("isspace:", s.isspace())
print("isalnum:", s.isalnum())

# round(x, n) avrunder til n desimaler
tall1 = 4.7
resultat1 = round(tall1, 5)
print("rounded:", resultat1)
print(round(3.1459, 2))  # -> 3.14

# format for strings with floats (brukes ofte ved f.eks. print)
val = 3.14159265
print(format(val, ".3f"))  # 3 desimaler
```

## Løkker
Kort tekst: continue fortsetter neste iterasjon, break avbryter løkken.

```
# Eksempel på continue og break i en for-løkke
for i in range(5):
    if i == 2:
        print("continue ved", i)
        continue
    if i == 4:
        print("break ved", i)
        break
    print("i =", i)
```

## Liste
Kort tekst: Opprette, oppdatere, hente ut elementer, og nyttige metoder for lister.

```
# Opprette en liste
a = ['banan', 'eple', 'appelsin', 'plomme']
print("liste a:", a)

# Random liste comprehension eksempel (bruker random)
import random as rnd
f = [rnd.randint(-100, 100) for _ in range(25)]
print("25 random tall:", f[:10], "...")

# Hente via index og oppdatere
print(a[0])  # skriver banan
a[0] = 'kiwi'  # endrer index 0 fra banan til kiwi
print("etter oppdatering:", a)

# Legge til elementer
a.append('kiwi')     # putter kiwi bakerst i listen
a.insert(1, 'sitron') # putter sitron på indeks 1
print("etter append/insert:", a)

# Lengde
print('len(a) =', len(a))

# Loop for å printe elementer under hverandre
for b in a:
    print(" -", b)

# Bytt fra streng av tall til int-liste
b = '5 3 3 5 2 6'
ab = [int(s) for s in b.split()]
print("konvertert:", ab)

# remove/pop/count/sort
a.pop(0)  # fjerner første element
print("etter pop(0):", a)
a.append('eple')
print("count av 'eple':", a.count('eple'))
a.sort()
print("sortert stigende:", a)
a.sort(reverse=True)
print("sortert synkende:", a)
```

## Random
Kort tekst: Bruke random for tilfeldigheter: randint, choice, shuffle.

```
import random as rnd
c = [1, 3, 5, 6]
print("randint:", rnd.randint(1, 100))
print("choice fra c:", rnd.choice(c))
print("choice inline:", rnd.choice([1,2,3,4]))

rnd.shuffle(c)
print("shuffled c:", c)
```

## Pandas
Kort tekst: Grunnleggende operasjoner med pandas: lese filer, utforske, velge data, filtrere, endre, og håndtere manglende verdier.

```
import pandas as pd
import numpy as np

# Eksempel: lage et lite DataFrame for demonstrasjon
df = pd.DataFrame({
    "etternavn": ["Hansen", "Olsen", "Nilsen", "Larsen"],
    "kjoenn": ["mann", "kvinne", "mann", "kvinne"],
    "fylke": ["Hordaland", "Oslo", "Hordaland", "Bergen"],
    "komiteer": ["Næringskomiteen", "Utdanning", "Næringskomiteen", "Helse"],
    "fodselsaar": [1975, 1988, 1992, 1980],
    "alder": [48, 35, 31, 43]
})
df.index.name = 'index'
df

# Hvis du har en fil (eksempel)
# df = pd.read_csv("stortingsrepresentanter.csv", sep=";", index_col=False)

# Utforske dataframe
print('dtypes:\n', df.dtypes)
print('\nhead():\n', df.head())
print('\nshape:', df.shape)
print('\ninfo():'); df.info()
print('\ndescribe():\n', df.describe())

# Velge kolonner
print('\ndf["etternavn"]:\n', df["etternavn"])
print('\ndf[["kjoenn","komiteer"]]:\n', df[["kjoenn", "komiteer"]])

# .loc og .iloc eksempler
print('\ndf.loc[0]:\n', df.loc[0])
print('\ndf.loc[0:2]:\n', df.loc[0:2])
print('\ndf.loc[(df.fylke=="Hordaland")]:\n', df.loc[(df.fylke=="Hordaland")])

# Flere filtre
print('\ndf.loc[(df.komiteer=="Næringskomiteen") & (df.fylke=="Hordaland")]:\n',
      df.loc[(df.komiteer=="Næringskomiteen") & (df.fylke=="Hordaland")])

# iloc eksempler
print('\ndf.iloc[0]:\n', df.iloc[0])
print('\ndf.iloc[[0,1]]:\n', df.iloc[[0,1]])
print('\ndf.iloc[1:3, 1:4]:\n', df.iloc[1:3, 1:4])

# Filter og sort
print('\nFilter alder > 30:\n', df[df['alder'] > 30])
print('\nSorter stigende etter alder:\n', df.sort_values('alder'))
print('\nSorter synkende:\n', df.sort_values('alder', ascending=False))

# Legge til/endre kolonner
df['by'] = ['Bergen', 'Oslo', 'Bergen', 'Bergen']
print('\nNy kolonne by:\n', df.head())

df.rename(columns={'etternavn': 'Etternavn'}, inplace=True)
print('\nRenamed columns:\n', df.columns)

# NaN eksempler
df2 = df.copy()
df2.loc[1, 'by'] = np.nan
print('\nHar NaN?\n', df2.isnull())
print('\nDropna:\n', df2.dropna())
print('\nFillna med 0:\n', df2.fillna(0))
```

## Object-oriented programming (OPP)
Kort tekst: Eksempler på klasser, instanser, arv, private attributter og inner classes.

```
# Enkel klasse og objekt
class Myclass:
    x = 5

print(Myclass)
ob1 = Myclass()
print("ob1.x =", ob1.x)

# Deleklasser og init
class Person2:
    def __init__(self, name, age=18):
        self.name = name
        self.age = age

k = Person2("Danni", 17)
print("k.name, k.age:", k.name, k.age)

# Metoder og __str__
class Person:
    def __init__(self, name, age=18):
        self.name = name
        self.age = age
    def greet(self):
        return "Hello, " + self.name
    def __str__(self):
        return f"{self.name} ({self.age})"

p1 = Person("Tobias", 36)
print(p1)
print(p1.greet())

# Privat attributt og tilgang
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

e = Employee("Linus", 50000)
print(e.name)
# Direkte tilgang (navnemangling)
print("salary via mangling:", e._Employee__salary)

# Inner class eksempel
class Outer:
    def __init__(self):
        self.name = "Outer"
    class Inner:
        def __init__(self, outer=None):
            self.outer = outer
        def display(self):
            if self.outer:
                print(f"Outer class name: {self.outer.name}")
            else:
                print("No outer instance provided")

outer = Outer()
inner = outer.Inner(outer)
inner.display()

# Practical Car example with inner Engine class
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        self.engine = self.Engine()

    class Engine:
        def __init__(self):
            self.status = "Off"
        def start(self):
            self.status = "Running"
            print("Engine started")
        def stop(self):
            self.status = "Off"
            print("Engine stopped")

    def drive(self):
        if self.engine.status == "Running":
            print(f"Driving the {self.brand} {self.model}")
        else:
            print("Start the engine first!")

car = Car("Toyota", "Corolla")
car.drive()
car.engine.start()
car.drive()
```
