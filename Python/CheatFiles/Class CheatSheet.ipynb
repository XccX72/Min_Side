# Classes Notes
Quick reference guide for Python classes and object-oriented programming concepts.

## Basic Class Example
Simple class creation and object instantiation.

```python
class MathExample:
    x = 6.7
    y = 69

p1 = MathExample()
p2 = MathExample()
p3 = MathExample()

del p1

print(p2.x, p3.x)
```

## Empty Class
A class can be defined with `pass` to create an empty structure.

```python
class EmptyClass:
    pass
```

## The __init__() Method
Used to initialize object properties when an object is created.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Emil", 36)
print(p1.name)
print(p1.age)
```

## The `self` Parameter
Represents the instance of the class; it doesn’t have to be named `self`.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("Emil", 25)
p1.greet()
```

## Multiple Objects
Each object can hold its own property values.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def print_name(self):
        print(self.name)

p1 = Person("Tobias")
p2 = Person("Linus")

p1.print_name()
p2.print_name()
```

## Example with Methods
Methods are functions defined inside a class.

```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.brand} {self.model}")

car1 = Car("Toyota", "Corolla", 2020)
car1.display_info()
```

## Class and Instance Properties
Class properties are shared by all instances; instance properties are unique per object.

```python
class Person:
    species = "Human"  # Class property

    def __init__(self, name):
        self.name = name  # Instance property

p1 = Person("Emil")
p2 = Person("Tobias")

print(p1.name)
print(p2.name)
print(p1.species)
print(p2.species)
```

## Modifying a Class Property
Class properties are changed using the class name.

```python
class Person:
    lastname = ""

    def __init__(self, name):
        self.name = name

p1 = Person("Linus")
p2 = Person("Emil")

Person.lastname = "Refsnes"

print(p1.lastname)
print(p2.lastname)
```

## Add Properties to an Object
You can add attributes dynamically to objects.

```python
class Person:
    def __init__(self, name):
        self.name = name

p1 = Person("Tobias")

p1.age = 25
p1.city = "Oslo"

print(p1.name)
print(p1.age)
print(p1.city)
```

## The __str__() Method
Controls how objects are represented as strings.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ({self.age})"

p1 = Person("Tobias", 36)
print(p1)
```

## Inheritance
Child classes can inherit properties and methods from parent classes.

```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def print_name(self):
        print(self.firstname, self.lastname)

class Student(Person):
    pass

x = Student("Mike", "Olsen")
x.print_name()
```

## Overriding __init__ with super()
The `super()` function calls the parent class’s constructor.

```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

class Student(Person):
    def __init__(self, fname, lname):
        super().__init__(fname, lname)

s = Student("John", "Doe")
s.print_name()
```

## Polymorphism
Different classes can define the same method name with different behavior.

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Move!")

class Car(Vehicle):
    pass

class Boat(Vehicle):
    def move(self):
        print("Sail!")

class Plane(Vehicle):
    def move(self):
        print("Fly!")

car1 = Car("Ford", "Mustang")
boat1 = Boat("Ibiza", "Touring 20")
plane1 = Plane("Boeing", "747")

for x in (car1, boat1, plane1):
    print(x.brand, x.model)
    x.move()
```

## Encapsulation
Protects internal object data and restricts direct access.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age  # Private property

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive")

p1 = Person("Tobias", 25)
print(p1.get_age())

p1.set_age(26)
print(p1.get_age())
```

## Protected Properties
__ to make private. Then you need .get and .set to change or get values

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.__age = age

  def get_age(self):
    return self.__age

p1 = Person("Tobias", 25)
print(p1.get_age())
```

A single underscore (`_`) indicates a protected property (by convention).

```python
class Person:
    def __init__(self, name, _salary):
        self.name = name
        self._salary = _salary  # Protected property

p1 = Person("Linus", 50000)
print(p1.name)
print(p1._salary)
```

## Inner Classes
A class defined inside another class. Used for grouping logic.

```python
class Outer:
    def __init__(self):
        self.name = "Outer"

    class Inner:
        def __init__(self):
            self.name = "Inner"

        def display(self):
            print("Hello from inner class")

outer = Outer()
inner = outer.Inner()
inner.display()
```

## Access Outer Class from Inner
The inner class can access the outer class by receiving it as a parameter.

```python
class Outer:
    def __init__(self):
        self.name = "Emil"

    class Inner:
        def __init__(self, outer):
            self.outer = outer

        def display(self):
            print(f"Outer class name: {self.outer.name}")

outer = Outer()
inner = outer.Inner(outer)
inner.display()
```

## Practical Inner Class Example
Inner classes can represent components within a larger system.

```python
class Computer:
    def __init__(self):
        self.cpu = self.CPU()
        self.ram = self.RAM()

    class CPU:
        def process(self):
            print("Processing data...")

    class RAM:
        def store(self):
            print("Storing data...")

computer = Computer()
computer.cpu.process()
computer.ram.store()
```
