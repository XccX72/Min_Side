# Pandas + Matplotlib – Komplett guide med forklaring før hvert eksempel

## Importere biblioteker

Vi starter med å importere pandas og matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

### pd.read_csv()
Leser en CSV-fil og returnerer en DataFrame.

```python
df_csv = pd.read_csv('data.csv', encoding='utf-8')
df_csv.head()
```

### pd.read_json()
Leser JSON-data til en DataFrame.

```python
df_json = pd.read_json('data.json')
df_json.head()
```

### pd.read_excel()
Leser et Excel-ark.

```python
df_xlsx = pd.read_excel('fil.xlsx', sheet_name='Ark1')
df_xlsx.head()
```

Filtrerer rader med SQL‑lignende syntaks.

```python
df_csv.query('alder >= 18 and kjonn == "K"')
```

Sorter DataFrame etter kolonner.

```python
df_csv.sort_values(['alder','navn'], ascending=[False,True])
```

### df.sample()
Tilfeldig utvalg av rader.

```python
df_csv.sample(n=5, random_state=1)
```

### df.rename()
Gir nye kolonnenavn.

```python
df_csv.rename(columns={'old':'ny'})
```

### df.assign()
Legger til beregnet kolonne.

```python
df_csv.assign(bmi=df_csv['vekt']/(df_csv['hoyde']/100)**2)
```

### df.drop()
Fjerner kolonner eller rader.

```python
df_csv.drop(columns=['uønsket'])
```

### df.astype()
Endrer kolonnedatatype.

```python
df_csv['alder'] = df_csv['alder'].astype('int64')
```

### df.set_index()
Setter en kolonne som indeks.

```python
df_csv.set_index('id', inplace=False)
```

### df.reset_index()
Nullstiller indeks.

```python
df_csv.reset_index(drop=True)
```

### df.duplicated()
Finner duplikater.

```python
df_csv[df_csv.duplicated(subset=['id'], keep=False)]
```

### df.isna()
Sjekker manglende verdier.

```python
df_csv.isna().sum()
```

### df.fillna()
Fyller inn manglende verdier.

```python
df_csv['poeng'] = df_csv['poeng'].fillna(0)
```

### df.dropna()
Fjerner rader/kolonner med NaN.

```python
df_csv.dropna(subset=['navn','alder'])
```

### df.value_counts()
Viser hvor mange ganger en verdi vises fra en kolonne fra et datasett

```python
df_csv['by'].value_counts()
```

### df.nunique()
Antall unike verdier.

```python
df_csv['by'].nunique()
```

### df.groupby().agg()
Gruppert statistikk.

```python
df_csv.groupby('by')['poeng'].agg(['mean','max'])
```

```python
###df["kolonne"].value
Gjør at du bare få navnene til dataen du leter ikke og alt det andre som index 
```

### pd.pivot_table()
Lager pivot‑tabell.

```python
pd.pivot_table(df_csv, index='by', columns='aar', values='poeng', aggfunc='mean')
```

### pd.concat()
Slår sammen DataFrames.

```python
pd.concat([df_csv, df_csv], ignore_index=True)
```

### pd.merge()
Join mellom DataFrames.

```python
pd.merge(df_csv, df_csv, on='id', how='left')
```

### pd.to_datetime()
Konverterer til datetime.

```python
df_csv['dato'] = pd.to_datetime(df_csv['dato'])
```

### df['dato'].dt.year
Henter år fra datetime.

```python
df_csv['aar'] = df_csv['dato'].dt.year
```

### df['tekst'].str.lower()
Tekst‑funksjoner.

```python
df_csv['domene'] = df_csv['epost'].str.split('@').str[-1]
```

### Series.map()
Mappe verdier én‑til‑én.

```python
df_csv['k'] = df_csv['kode'].map({'M':'Mann','K':'Kvinne'})
```

### Series.apply()
Kjører funksjon på hver verdi.

```python
df_csv['len'] = df_csv['navn'].apply(len)
```

### DataFrame.apply()
Funksjon over rader/kolonner.

```python
df_csv['sum'] = df_csv[['a','b']].apply(sum, axis=1)
```

### DataFrame.pipe()
Kjeder operasjoner på ryddig måte.

```python
res = df_csv.pipe(lambda d: d.dropna()).pipe(lambda d: d.query('alder > 18'))
res.head()
```

### pd.melt()
Omformer bred → lang tabell.

```python
long = pd.melt(df_csv, id_vars='navn', var_name='aar', value_name='poeng')
long.head()
```

### df.pivot()
Omformer lang → bred tabell.

```python
wide = long.pivot(index='navn', columns='aar', values='poeng')
wide.head()
```

### df.plot()
Rask plotting med matplotlib.

```python
df_plot = df_csv.set_index('aar')['poeng']
df_plot.plot()
plt.xlabel('År')
plt.ylabel('Poeng')
plt.title('Poeng per år')
plt.show()
```
